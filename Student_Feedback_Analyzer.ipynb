Create our sample dataset

```python

# KEMU Student Feedback Analyzer
# Step 1: Create sample student feedback

feedback = [
    "The Wi-Fi is very slow in the library",
    "The registration process was easy and fast",
    "The food in the cafeteria is bad",
    "I am happy with the new student portal",
    "The Wi-Fi is not working properly",
    "The lecturers are helpful and supportive",
    "Registration is slow and frustrating",
    "The cafeteria food is good",
    "The student portal is difficult to use",
    "I love the improved internet connection"
]

for number, comment in enumerate(feedback, start=1):
    print(f"{number}. {comment}")
```

Clean the student feedback


```python

import re

# Step 2: Create a function to clean text
def clean_text(text):
    text = text.lower()  # Convert to lowercase
    text = re.sub(r"[^a-z0-9\s]", " ", text)  # Remove punctuation
    text = re.sub(r"\s+", " ", text).strip()  # Remove extra spaces
    return text

# Apply cleaning to all student comments
cleaned_feedback = []

for comment in feedback:
    cleaned = clean_text(comment)
    cleaned_feedback.append(cleaned)

# Display original and cleaned comments
for original, cleaned in zip(feedback, cleaned_feedback):
    print("Original:", original)
    print("Cleaned: ", cleaned)
    print("-" * 50)
```

Tokenization

```python

# Step 3: Tokenize the cleaned feedback

tokenized_feedback = []

for comment in cleaned_feedback:
    tokens = comment.split()
    tokenized_feedback.append(tokens)

# Display the results
for comment, tokens in zip(cleaned_feedback, tokenized_feedback):
    print("Comment:", comment)
    print("Tokens: ", tokens)
    print("-" * 50)
```

Remove stop words

```python

# Step 4: Stop-word removal

stop_words = {
    "the", "is", "a", "an", "in", "on", "at",
    "and", "are", "was", "were", "to", "of",
    "with", "very", "i", "am", "my"
}

# Keep important words such as "not"
filtered_feedback = []

for tokens in tokenized_feedback:
    filtered_tokens = [
        word for word in tokens
        if word not in stop_words
    ]
    filtered_feedback.append(filtered_tokens)

# Compare tokens before and after removal
for original_tokens, filtered_tokens in zip(
    tokenized_feedback, filtered_feedback
):
    print("Before:", original_tokens)
    print("After: ", filtered_tokens)
    print("-" * 60)
```

Combine the preprocessing stages

```python

import re
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

stop_words = {
    "the", "is", "a", "an", "in", "on", "at",
    "and", "are", "was", "were", "to", "of",
    "with", "very", "i", "am", "my"
}

def preprocess_text(text):
    # 1. Convert to lowercase
    text = text.lower()

    # 2. Remove punctuation
    text = re.sub(r"[^a-z0-9\s]", " ", text)

    # 3. Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    # 4. Tokenize
    tokens = text.split()

    # 5. Remove selected stop words
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    # 6. Lemmatize words as verbs
    tokens = [
        lemmatizer.lemmatize(word, pos="v")
        for word in tokens
    ]

    return tokens

# Test the complete pipeline
sample = "The Wi-Fi is NOT working properly!!!"

print("Original:", sample)
print("Processed:", preprocess_text(sample))
```

Prepare feedback for sentiment analysis

```python

import pandas as pd

data = {
    "feedback": [
        "The Wi-Fi is very slow in the library",
        "The registration process was easy and fast",
        "The food in the cafeteria is bad",
        "I am happy with the new student portal",
        "The Wi-Fi is not working properly",
        "The lecturers are helpful and supportive",
        "Registration is slow and frustrating",
        "The cafeteria food is good",
        "The student portal is difficult to use",
        "I love the improved internet connection"
    ],
    "sentiment": [
        "Negative",
        "Positive",
        "Negative",
        "Positive",
        "Negative",
        "Positive",
        "Negative",
        "Positive",
        "Negative",
        "Positive"
    ]
}

df = pd.DataFrame(data)

df["processed_text"] = df["feedback"].apply(
    lambda text: " ".join(preprocess_text(text))
)

df
```

Bag of Words

```python

# Generate a Bag of Words Word Cloud from Student Feedback

!pip -q install wordcloud

import matplotlib.pyplot as plt
from wordcloud import WordCloud
from collections import Counter

# Combine the processed student feedback into one text
all_text = " ".join(df["processed_text"].astype(str))

# Count how often each word appears
word_counts = Counter(all_text.split())

# Display the 15 most frequent words
print("Most common words:")
for word, count in word_counts.most_common(15):
    print(f"{word}: {count}")

# Generate the word cloud
wordcloud = WordCloud(
    width=1200,
    height=700,
    background_color="white",
    colormap="viridis",
    max_words=100,
    collocations=False
).generate_from_frequencies(word_counts)

# Display the visualization
plt.figure(figsize=(15, 8))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title(
    "Most Common Words in Student Feedback\n(After Text Preprocessing)",
    fontsize=18,
    pad=20
)
plt.show()
```

Explore the results

```python

import matplotlib.pyplot as plt

sentiment_counts = df["sentiment"].value_counts()

print(sentiment_counts)

sentiment_counts.plot(kind="bar")
plt.title("Student Feedback: Label Counts")
plt.xlabel("Sentiment")
plt.ylabel("Number of Comments")
plt.xticks(rotation=0)
plt.show()
```

Build a basic sentiment classifier

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

model = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("classifier", LogisticRegression(max_iter=1000))
])

# Train on all ten examples for this demonstration
model.fit(df["processed_text"], df["sentiment"])

new_comments = [
    "The internet is slow and frustrating",
    "The new portal is helpful and easy to use",
    "The cafeteria food is bad"
]

predictions = model.predict([
    " ".join(preprocess_text(comment))
    for comment in new_comments
])

for comment, prediction in zip(new_comments, predictions):
    print("Feedback:", comment)
    print("Predicted sentiment:", prediction)
    print("-" * 45)
```
